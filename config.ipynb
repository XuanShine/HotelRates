```python
import yaml
import pandas as pd
import numpy as np
import gspread
from random import choice
from gspread_dataframe import get_as_dataframe, set_with_dataframe
from sheet import spreadsheet as ss
import xotelo
from datetime import date
pd.set_option('future.no_silent_downcasting', True)
import os
# Get the folder where this script is located
# base_dir = os.path.dirname(os.path.abspath(__file__))
base_dir = os.path.dirname(os.path.abspath("."))

yaml_config_path = os.path.join(base_dir, "HOTELRATES", 'hotels.yml')

with open(yaml_config_path, "r") as f_in:
    HOTELS = yaml.safe_load(f_in)
SHORT_TO_ID = {
    hotel['short']: hotel['id_tripadvisor'] 
    for hotel in HOTELS.values() 
}
SHORT_NAMES = {data['short'] for data in HOTELS.values()}
WORKSHEET = ss.client.open("Prix Aroma").worksheet("config")

def load_config():
    df = get_as_dataframe(WORKSHEET, index_col=0, skip_blank_lines=True, evaluate_formulas=True)
    df['value'] = df['value'].astype('object')
    return df

def reset_update_go(config):
    config.loc["update_go", "value"] = False

def export_config(df):
    set_with_dataframe(
        WORKSHEET,
        df,
        include_index=True
    )
```
