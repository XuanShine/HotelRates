```python
import yaml
import pandas as pd
import numpy as np
import gspread
from random import choice
from gspread_dataframe import get_as_dataframe, set_with_dataframe
from sheet import spreadsheet as ss
import xotelo
from datetime import date
pd.set_option('future.no_silent_downcasting', True)
import os
WORKSHEET = ss.client.open("Prix Aroma").worksheet("Feuille1")

def load_df():
    df = get_as_dataframe(WORKSHEET, index_col=0, parse_dates=True, dayfirst=True, skip_blank_lines=True, evaluate_formulas=True)
    return df

def clean_past(df):
    df = df[pd.Timestamp.now().date():]
    return df

def export_all_df(df):
    df_export = df.copy()
    df_export.index = df_export.index.strftime('%d/%m/%Y')
    set_with_dataframe(
        WORKSHEET, 
        df_export, 
        include_index=True,
        resize=True
    )
```
